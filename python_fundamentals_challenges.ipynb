## 1. Capitalize words separated by spaces

Preserve spaces and capitalize the first character after a space. Return an empty string for empty input.

```python
def capitalize_words(input_string):
    if not input_string:
        return ""
    chars = list(input_string)
    for index, char in enumerate(chars):
        if index == 0 or chars[index - 1] == " ":
            chars[index] = char.upper()
    return "".join(chars)

print(capitalize_words("hello world"))
print(capitalize_words("python programming is fun"))
assert capitalize_words("") == ""
assert capitalize_words("hello  world") == "Hello  World"
```

## 2. Fizz on numbers divisible by a specified divisor

Return strings for **all** elements. Reject zero divisors with an explanatory exception.

```python
def fizz_buzz(start, end, divisor):
    if divisor == 0:
        raise ValueError("divisor must not be zero")
    result = []
    for number in range(start, end + 1):
        if number % divisor == 0:
            result.append("Fizz")
        else:
            result.append(str(number))
    return result

print(fizz_buzz(1, 10, 3))
print(fizz_buzz(5, 15, 5))
assert fizz_buzz(1, 5, 3) == ["1", "2", "Fizz", "4", "5"]
assert fizz_buzz(5, 4, 2) == []
```

## 3. Collect unique elements from nested lists

```python
def unique_elements(nested_lists):
    unique = set()
    for group in nested_lists:
        for element in group:
            unique.add(element)
    return unique

print(unique_elements([[1, 2, 3], [2, 3, 4], [4, 5, 6]]))
assert unique_elements([[10, 20], [20, 30, 40], [10, 50]]) == {10, 20, 30, 40, 50}
assert unique_elements([]) == set()
```

## 4. Count case-insensitive words, ignoring punctuation

Treat punctuation as separators, normalize case, and split on arbitrary whitespace. This is a simple English-language tokenizer, not full NLP.

```python
import string

def word_frequency(input_string):
    punctuation_to_spaces = str.maketrans({char: " " for char in string.punctuation})
    words = input_string.lower().translate(punctuation_to_spaces).split()
    frequencies = {}
    for word in words:
        frequencies[word] = frequencies.get(word, 0) + 1
    return frequencies

print(word_frequency("Hello world! Hello everyone."))
print(word_frequency("Python programming is fun. Programming in Python is exciting."))
assert word_frequency("Hi, hi!   BYE?\nbye") == {"hi": 2, "bye": 2}
assert word_frequency("") == {}
```
